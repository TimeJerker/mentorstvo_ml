```python
import csv
import json

class lab1:
    def __init__(self, path_file):
        self.path_file = path_file
        self.data = []
        self.load_data()

    def load_data(self):
        with open(self.path_file, "r") as file:
            reader = csv.DictReader(file)
            
            for row in reader:
                has_empty = False
                for value in row.values():
                    if value.strip() == "":
                        has_empty = True
                        break
            
                if has_empty != True:
                    row["age"] = int(row["age"])
                    row["bmi"] = float(row["bmi"])
                    row["charges"] = float(row["charges"])
                    self.data.append(row)

    def __get_column(self, column_name) -> list:
        return [row[column_name] for row in self.data]

    def calc_mean(self,column_name):
        arg = self.__get_column(column_name)
        if not arg:
            return 0.0
        return sum(arg) / len(arg)

    def calc_disper(self, column_name) -> float:
        arg = self.__get_column(column_name)
        mean_arg = self.calc_mean(column_name)

        squared_diffs = [(x - mean_arg)**2 for x in arg]

        return sum(squared_diffs) / len(squared_diffs)

    def prob_smoker_if_high_charges(self, threshold: int = 15000) -> float:
        count_B = []
        count_A = [] 
        for row in self.data:
            if row["charges"] > threshold:
                count_B.append(row)

        for row in count_B:
            if row["smoker"] == "yes":
                count_A.append(row)

        return len(count_A) / len(count_B)

analyze = lab1("insurance.csv")

report = {
    "mean_charges": analyze.calc_mean("charges"),
    "disper_charges": analyze.calc_disper("charges"),
    "prob_smoker_if_high_charges": analyze.prob_smoker_if_high_charges(threshold = 20000)
}

with open("report.json", "w") as f:
    json.dump(report, f, indent = 4)
```
